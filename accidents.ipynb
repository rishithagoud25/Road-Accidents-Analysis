```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("indian_roads_dataset.csv")


print("Missing values before cleaning:")
print(df['festival'].isnull().sum()) #Check null values in festival column

df['festival'] = df['festival'].fillna('Normal Day') #Replace null values with 'Normal Day'

df['date'] = pd.to_datetime(df['date']) #Converting date column into proper date format

print("Missing values after cleaning:")
print(df['festival'].isnull().sum())
```

```python
plt.figure(figsize=(10,5))


sns.countplot(data=df, x='hour', palette= "tab10")


plt.title("Accidents by Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Accidents")

plt.savefig("hourly_accidents.png")

plt.show()
```

```python
plt.figure(figsize=(10,5))

weather_counts = df['weather'].value_counts()
plt.pie(weather_counts, labels=weather_counts.index, autopct='%1.1f%%')

plt.title("Weather Analysis")
plt.savefig("weather_analysis.png")

plt.show()
```

```python
plt.figure(figsize=(10,5))

cause_counts = df['cause'].value_counts()

plt.pie(cause_counts, labels=cause_counts.index, autopct='%1.2f%%')
plt.title("Causes of Road Accidents")
plt.savefig("accident_causes.png")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
#State-wise accident analysis
plt.figure(figsize=(10,5))

state_counts = df['state'].value_counts()

sns.barplot(x=state_counts.index, y=state_counts.values, palette='viridis')

plt.title("Accidents by State")
plt.xlabel("State")
plt.ylabel("Number of Accidents")

plt.xticks(rotation=15)

plt.savefig("state_accidents.png")
plt.show()

#City-wise accident analysis
city_severity = pd.crosstab(df['city'], df['accident_severity'])

city_severity.plot(kind='bar')

plt.title("City vs Accident Severity")
plt.xlabel("City")
plt.ylabel("Number of Accidents")

plt.xticks(rotation=20)

plt.savefig("city_severity.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['risk_score'], bins=20)
plt.title('Distribution of Risk Score')
plt.xlabel('Risk Score')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['casualties'], bins=15)
plt.title('Distribution of Casualties')
plt.xlabel('Casualties')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['vehicles_involved'], df['casualties'])
plt.title('Vehicles Involved vs Casualties')
plt.xlabel('Vehicles Involved')
plt.ylabel('Casualties')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['temperature'], df['risk_score'])
plt.title('Temperature vs Risk Score')
plt.xlabel('Temperature')
plt.ylabel('Risk Score')
plt.show()
```

```python
hourly = df.groupby('hour').size()

plt.figure(figsize=(10,5))
plt.plot(hourly.index, hourly.values, marker='o')
plt.title('Number of Accidents by Hour')
plt.xlabel('Hour')
plt.ylabel('Number of Accidents')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='weather', data=df)
plt.title('Weather Conditions')
plt.xticks(rotation=45)
plt.show()
```

```python
print("Accident Analysis Report")

print("Total Accidents in the data(2022-25):", len(df))

print("Total Casualties:", df['casualties'].sum())

print("Top 3 Dangerous States:")
print(df['state'].value_counts().head(3))

print("Top 3 Dangerous Cities:")
print(df['city'].value_counts().head(3))

print("Most Common Cause:", df['cause'].mode()[0])

print("Peak Accident Hour:", df['hour'].mode()[0])
```
